```python
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

category_analysis["Profit Margin"] = (
    category_analysis["Profit"] / category_analysis["Sales"]
) * 100

category_analysis
```

```python
monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales
```

```python
monthly_sales.idxmax(),monthly_sales.max()
```

```python
monthly_sales.idxmin(),monthly_sales.min()
```

```python
region_analysis = df.groupby("Region")[["Sales","Profit"]].sum()
region_analysis
```

```python
region_analysis.idxmax(),monthly_sales.max()
```

```python
region_analysis.idxmin(),monthly_sales.min()
```

```python
top_cust = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_cust
```

```python
cust_prof = df.groupby("Customer Name")["Profit"].sum().sort_values(ascending=False).head(10)
cust_prof
```

```python
loss_cust = df.groupby("Customer Name")["Profit"].sum().sort_values().head(10)
loss_cust
```

```python
top_prod = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(10)
top_prod
```

```python
loss_prod = df.groupby("Product Name")["Profit"].sum().sort_values().head(10)
loss_prod
```

```python
discount_analysis = df.groupby("Category")["Discount"].mean().sort_values(ascending=False)
discount_analysis
```

```python
disc_prof = df.groupby("Discount")["Profit"].agg(["sum","mean","count"])
disc_prof
```
